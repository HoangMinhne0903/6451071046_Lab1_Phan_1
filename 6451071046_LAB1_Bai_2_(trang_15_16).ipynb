```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

df = pd.read_csv('titanic.csv')
df.columns = df.columns.str.lower()

# Tổng số người thân đi cùng = sibsp + parch
df['relatives'] = df['sibsp'] + df['parch']

print("Thống kê số lượng hành khách theo số người đi cùng:")
print(df['relatives'].value_counts().sort_index())

print("\nTỉ lệ sống sót trung bình theo số người đi cùng:")
print(df.groupby('relatives')['survived'].mean())

plt.figure(figsize=(10, 5))
sns.pointplot(x='relatives', y='survived', data=df, color='teal')
plt.title('Tỉ lệ sống sót theo Tổng số người thân đi cùng (Relatives)')
plt.xlabel('Số người đi cùng (SibSp + Parch)')
plt.ylabel('Tỉ lệ sống sót (Survived)')
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```

```python
import numpy as np
import pandas as pd

print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

df = pd.read_csv('titanic.csv')
df.columns = df.columns.str.lower()

# 1. Xóa các thuộc tính không cần thiết hoặc thiếu quá nhiều dữ liệu
cols_to_drop = [
    'passengerid',
    'ticket',
    'cabin',
    'boat',
    'body',
    'home.dest',
]
df.drop(
    columns=[col for col in cols_to_drop if col in df.columns], inplace=True
)

# 2. Điền dữ liệu thiếu cho cột 'age' bằng giá trị ngẫu nhiên quanh (mean - std, mean + std)
mean_age = df['age'].mean()
std_age = df['age'].std()
null_age_count = df['age'].isnull().sum()
random_ages = np.random.randint(
    mean_age - std_age, mean_age + std_age, size=null_age_count
)
df.loc[df['age'].isnull(), 'age'] = random_ages

# 3. Điền dữ liệu thiếu cho 'embarked' bằng giá trị phổ biến nhất (mode)
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

# 4. Điền dữ liệu thiếu cho 'fare' bằng trung vị (median)
df['fare'] = df['fare'].fillna(df['fare'].median())

print("Kiểm tra lại dữ liệu thiếu sau tiền xử lý:")
print(df.isnull().sum())
print("\nCác cột còn lại trong dữ liệu:")
print(df.columns.tolist())
```

**Câu 7:**

```python
import pandas as pd

print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

df = pd.read_csv('titanic.csv')
df.columns = df.columns.str.lower()

# Làm sạch giá trị thiếu trước khi biến đổi
df['age'] = df['age'].fillna(df['age'].median())
raw_fare = df['fare'].fillna(df['fare'].median())
df['embarked'] = df['embarked'].fillna(df['embarked'].mode()[0])

# 1. Chuyển đổi thuộc tính giá vé (fare) về kiểu nguyên
df['fare'] = raw_fare.astype(int)

# 2. Trích xuất thuộc tính danh hiệu (title) từ tên (name)
df['title'] = df['name'].str.extract(' ([A-Za-z]+)\.', expand=False)
df['title'] = df['title'].replace(
    [
        'Lady',
        'Countess',
        'Capt',
        'Col',
        'Don',
        'Dr',
        'Major',
        'Rev',
        'Sir',
        'Jonkheer',
        'Dona',
    ],
    'Rare',
)
df['title'] = df['title'].replace({'Mlle': 'Miss', 'Ms': 'Miss', 'Mme': 'Mrs'})
title_mapping = {'Mr': 1, 'Miss': 2, 'Mrs': 3, 'Master': 4, 'Rare': 5}
df['title'] = df['title'].map(title_mapping).fillna(0).astype(int)

# 3. Chuyển đổi thuộc tính giới tính (sex) và cổng lên tàu (embarked) thành số
df['sex'] = df['sex'].map({'male': 0, 'female': 1}).astype(int)
df['embarked'] = df['embarked'].map({'S': 0, 'C': 1, 'Q': 2}).astype(int)

# 4. Chia giá trị thuộc tính tuổi (age) thành 8 nhóm theo các mốc gợi ý
age_bins = [-1, 11, 18, 22, 27, 33, 40, 66, 150]
age_labels = [0, 1, 2, 3, 4, 5, 6, 7]
df['age_group'] = pd.cut(df['age'], bins=age_bins, labels=age_labels).astype(
    int
)

# 5. Chia giá trị thuộc tính giá vé (fare) thành 6 nhóm theo các mốc gợi ý
fare_bins = [-1, 7.91, 14.454, 31, 99, 250, 1000]
fare_labels = [0, 1, 2, 3, 4, 5]
df['fare_group'] = pd.cut(raw_fare, bins=fare_bins, labels=fare_labels).astype(
    int
)

print("Kết quả 10 dòng đầu sau khi biến đổi các thuộc tính:")
print(
    df[[
        'sex',
        'embarked',
        'title',
        'age',
        'age_group',
        'fare',
        'fare_group',
    ]].head(10)
)
```

```python
import pandas as pd

print("Họ và tên: Nguyễn Lê Hoàng Minh - MSSV: 6451071046")

df = pd.read_csv('titanic.csv')
df.columns = df.columns.str.lower()

# Tiền xử lý giá trị thiếu
df['age'] = df['age'].fillna(df['age'].median())
df['fare'] = df['fare'].fillna(df['fare'].median())

# Tính số người đi cùng (relatives = sibsp + parch)
df['relatives'] = df['sibsp'] + df['parch']

# 1. Thuộc tính tuổi * loại vé (age_class)
df['age_class'] = df['age'] * df['pclass']

# 2. Thuộc tính giá vé trên mỗi người (fare_per_person) có làm trơn + 1
df['fare_per_person'] = df['fare'] / (df['relatives'] + 1)

print("Kết quả 10 dòng đầu với 2 thuộc tính mới:")
print(
    df[[
        'age',
        'pclass',
        'age_class',
        'fare',
        'relatives',
        'fare_per_person',
    ]].head(10)
)
```
